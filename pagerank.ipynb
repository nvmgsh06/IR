```python
import networkx as nx

# Step 1: Create a directed graph
G = nx.DiGraph()

# Step 2: Add edges between pages
G.add_edges_from([
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'C')
])

# Step 3: Apply PageRank algorithm
pagerank = nx.pagerank(G, alpha=0.85)

# Step 4: Display PageRank values
print("PageRank Scores:")

for page, score in pagerank.items():
    print(page, ":", round(score, 4))

# Step 5: Find the highest ranked page
ranked_page = max(pagerank, key=pagerank.get)

print("\nHighest Ranked Page:", ranked_page)
```
