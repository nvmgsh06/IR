```python
# Program for Document Retrieval using Inverted Files

# Sample documents
documents = {
    1: "Python is a programming language",
    2: "Python is used in data science",
    3: "Data science uses machine learning"
}

# Create inverted index
inverted_index = {}

for doc_id, text in documents.items():
    words = text.lower().split()

    for word in words:
        if word not in inverted_index:
            inverted_index[word] = []

        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)

# Display inverted index
print("Inverted Index:")
for word, doc_ids in inverted_index.items():
    print(word, ":", doc_ids)

# Search document
query = input("\nEnter search word: ").lower()

print("\nRetrieved Documents:")

if query in inverted_index:
    for doc_id in inverted_index[query]:
        print("Document", doc_id, ":", documents[doc_id])
else:
    print("No document found")
```
